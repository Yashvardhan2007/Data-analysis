```python
import pandas as pd
import os

# 1. Load the data
raw_data_dir = "../data/raw"
df_trader = pd.read_csv(os.path.join(raw_data_dir, 'historical_data.csv'))
df_sentiment = pd.read_csv(os.path.join(raw_data_dir, 'fear_greed_index.csv'))

# 2. Clean the dates in the Trader Data
# Hyperliquid timestamps are typically in milliseconds, let's convert it to a standard datetime
df_trader['Datetime'] = pd.to_datetime(df_trader['Timestamp'], unit='ms', errors='coerce')

# If the timestamp conversion fails or looks weird, fallback to parsing 'Timestamp IST'
if df_trader['Datetime'].isnull().all():
    df_trader['Datetime'] = pd.to_datetime(df_trader['Timestamp IST'], errors='coerce')

# Extract just the YYYY-MM-DD format to match the sentiment dataset
df_trader['date_key'] = df_trader['Datetime'].dt.strftime('%Y-%m-%d')

# 3. Clean the dates in the Sentiment Data
df_sentiment['date_key'] = pd.to_datetime(df_sentiment['date']).dt.strftime('%Y-%m-%d')

# 4. Merge the datasets together
df_merged = pd.merge(df_trader, df_sentiment, on='date_key', how='inner')

# 5. Show the results
print(f"Original Trader Rows: {len(df_trader)}")
print(f"Merged Data Rows: {len(df_merged)}")
print("\nFirst 3 rows of the successfully merged data:")
print(df_merged[['date_key', 'Account', 'Coin', 'Closed PnL', 'value', 'classification']].head(3))
```
